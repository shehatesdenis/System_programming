# Лабораторная работа №6

**Тема** — использование внешних библиотек в ассемблере.

**Вариант 23** — движение №5, красный и чёрный цвета следа, клавиша `l` для выхода, клавиша `p` для смены скорости.

[Условие и генератор вариантов](https://github.com/junaart/System_programming/blob/main/Lab_6/Lab_6.ipynb)


## Алгоритм

Движение начинается в левом верхнем углу. Курсор идёт вниз, вправо, вверх и влево, каждый раз исключая из рабочей области пройденную сторону. Так получается спираль, которая заполняет окно от краёв к центру. Когда все клетки пройдены, программа возвращается в левый верхний угол и повторяет проход другим цветом.

Каждая клетка отмечается точкой на красном или чёрном фоне с помощью цветовых пар `ncurses`. Во время ожидания следующего шага `wgetch` обрабатывает клавиши. Клавиша `p` переключает задержку между 35 и 8 мс, а `l` завершает программу и восстанавливает режим терминала. При изменении размеров окна спираль строится заново в новых границах.


## Сборка и запуск

Для сборки под Linux x86-64 нужны FASM, GCC и библиотека `ncurses` (в Debian и Ubuntu заголовки и файлы для линковки предоставляет пакет `libncurses-dev`). Из корня репозитория:

```sh
cd Lab_6
fasm task.asm task.o
gcc -no-pie task.o -lncurses -o task
./task
```

Нажми `p`, чтобы сменить скорость, и `l`, чтобы выйти.


## Проверка

В псевдотерминале размером 3 строки на 4 столбца проверены порядок обхода всех 12 клеток, смена красного фона на чёрный, изменение задержки по `p`, перезапуск после изменения размера окна и восстановление терминала при выходе по `l`. Отдельно проверены размеры 1×4, 4×1 и 1×1.
