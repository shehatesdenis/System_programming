## Задача 5. Раскачивающийся порядок символов

Для исходного файла и позиции `k` записать символы в порядке `k, k+1, k-1, k+2, k-2, …, k+m, k-m`. Позиции в файле считаются с единицы. Если позиция меньше 1 или выходит за конец файла, соответствующий символ пропускается.

**Алгоритм:** сначала прочитать символ в позиции `k`. Затем для каждого `j` от 1 до `m` прочитать позиции `k+j` и `k-j` и записать найденные символы в выходной файл. Доступ к позиции выполняется системным вызовом `lseek`, чтение и запись — вызовами `read` и `write`.

**Запуск:** `./task5 input.txt output.txt k m`

Исходный код: `task5.asm`.


```python
text = '123456789'
k, m = 5, 4
positions = [k] + [p for j in range(1, m + 1) for p in (k + j, k - j)]
''.join(text[p - 1] for p in positions if 1 <= p <= len(text))
```

## Задача 2. Каждый k-й символ

Прочитать из заданного текстового файла каждый `k`-й символ и записать его во второй файл. Позиции считаются от начала файла с единицы: выбираются символы на позициях `k, 2k, 3k, …`. Переводы строк также считаются символами файла.

**Алгоритм:** читать файл по одному байту, считать позицию, и записывать байт, когда номер позиции кратен `k`.

**Запуск:** `./task2 input.txt output.txt k`

Исходный код: `task2.asm`.


```python
text = 'abcdefghij'
k = 3
text[k - 1::k]
```
