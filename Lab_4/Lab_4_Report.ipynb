## Задача 13

Для заданного `n` найти числа от 10 до `n`, которые делятся на каждую из двух последних цифр (десятки и единицы). Если одна из цифр равна нулю, число не подходит: деление на ноль не определено.

**Алгоритм:** перебрать `k` от 10 до `n`; получить цифру единиц `k % 10` и цифру десятков `(k / 10) % 10`; если обе цифры ненулевые и `k` делится на обе, вывести `k`.

Исходный код находится в файлах `task13.asm` и `task13.c`. Ассемблерная версия читает строку через `read`, разбирает целое число и выводит найденные значения через `write`.


```python
# Контрольный пример на Python для демонстрации результата
n = 100
answer = [k for k in range(10, n + 1)
          if k % 10 != 0 and (k // 10) % 10 != 0
          and k % (k % 10) == 0 and k % ((k // 10) % 10) == 0]
answer
```

## Задача 5

Для заданного `n` посчитать количество целых чисел от 1 до `n`, которые не делятся ни на 11, ни на 5.

**Алгоритм:** перебрать все `k` от 1 до `n`; увеличить счётчик, если `k % 11 != 0` и `k % 5 != 0`.

Исходный код: `task5.asm` и `task5.c`. В FASM ввод выполнен системным вызовом `read`.


```python
n = 30
count = sum(1 for k in range(1, n + 1) if k % 11 != 0 and k % 5 != 0)
count
```

## Задача 2

Вычислить сумму `1 - 2² + 3² - 4² + … + (-1)ⁿ⁻¹ n²`.

**Алгоритм:** перебрать `k` от 1 до `n`; прибавлять `k²` при нечётном `k` и вычитать `k²` при чётном `k`. Для `n ≤ 0` сумма равна нулю.

Исходный код: `task2.asm` и `task2.c`.


```python
n = 5
sum((1 if k % 2 else -1) * k * k for k in range(1, n + 1))
```
